# Chapter 1 - Lab 2: Knowledge cutoff and the private-data gap

**Goal**: show *why* a plain LLM is not enough for customer care.

An LLM only knows what was in its training data. That creates two gaps:

1. **Time gap (knowledge cutoff)**: it does not know recent events or the latest regulations.
2. **Access gap (private data)**: it has never seen *your* orders, tickets, or policies.

Both gaps are closed later by giving the model **tools**, which turns it into an agent (Lab 3).

*Finance version*: "Give me the latest news about NVIDIA" (answer frozen at the cutoff).
*Customer-care version*: recent consumer-protection news, plus a private order lookup.

## 1. The time gap

```python
from openai import OpenAI
client = OpenAI()

response = client.responses.create(
    model="gpt-4.1",
    input="Give me the latest news about consumer-protection actions and customer-service "
          "complaints involving Indian e-commerce companies.",
)
print(response.output_text)
```

GPT-4.1's knowledge cutoff is mid-2024. Look for phrases like *"as of my last update"*: the model
answers with old news and cannot know this month's regulator actions or complaints.

## 2. The access gap

```python
response = client.responses.create(
    model="gpt-4.1",
    input="I am a NovaMart customer. What is the delivery status of my order ORD-78120?",
)
print(response.output_text)
```

The model either says it has no access, or (worse) **invents** a plausible answer. An invented
delivery date given to a real customer is a service failure. A customer-care agent must be
grounded in the company's own systems: order management, CRM, knowledge base and policies.
